```python
# yi = 3xi + 8 + N(0,1/2)
import numpy as np

# Generate x values
x = np.random.uniform(0, 1000, 5000)

# Generate noise ~ N(0, 1/2)
noise = np.random.normal(0, np.sqrt(1/2), 5000)

# Compute y values
y = 3 * x + 8 + noise

print(y)
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Generate data
x = np.random.uniform(0, 1000, 5000)
noise = np.random.normal(0,50, 5000)
y = 3 * x + 8 + noise

# Create scatter plot
plt.figure()
plt.scatter(x, y, s=1)  # s controls point size

# Labels and title
plt.xlabel("x")
plt.ylabel("y")
plt.title("Scatter plot of y = 3x + 8 + noise")

# Show plot
plt.show()
```
